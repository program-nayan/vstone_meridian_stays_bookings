```python
%sql
-- 1. Financial Mask (Amount)
CREATE OR REPLACE FUNCTION meridian_t12.gov.mask_amount(amount DECIMAL(10,2))
RETURN CASE
    WHEN is_account_group_member('t12_admins') OR is_account_group_member('t12_pii_readers') THEN amount
    ELSE CAST(ROUND(amount, 0) AS DECIMAL(10,2)) -- Partial masking / Redaction
END;
```

```python
%sql
-- 2. Email Mask
CREATE OR REPLACE FUNCTION meridian_t12.gov.mask_email(email STRING)
RETURN CASE
    WHEN is_account_group_member('t12_admins') OR is_account_group_member('t12_pii_readers') THEN email
    ELSE '[REDACTED EMAIL]'
END;
```

```python
%sql

-- 3. Name Mask
CREATE OR REPLACE FUNCTION meridian_t12.gov.mask_name(name STRING)
RETURN CASE
    WHEN is_account_group_member('t12_admins') OR is_account_group_member('t12_pii_readers') THEN name
    ELSE 'CONFIDENTIAL GUEST'
END;
```
