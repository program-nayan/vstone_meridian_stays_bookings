```python
%sql
CREATE OR REPLACE FUNCTION meridian_t12.gov.row_filter_by_destination_country(destination_id BIGINT)
RETURN 
  is_account_group_member('t12_admins') OR
  EXISTS (
    SELECT 1 
    FROM meridian_t12.gold.gold_dim_destinations d
    JOIN meridian_t12.gov.group_country_entitlements e 
      ON e.allowed_country = 'ALL' OR e.allowed_country = d.country
    WHERE d.destination_id = row_filter_by_destination_country.destination_id
      AND is_account_group_member(e.group_name)
  );
```
