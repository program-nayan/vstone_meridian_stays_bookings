```sql
SELECT 
    t.catalog_name AS catalog_name,
    t.schema_name AS schema_name,
    t.table_name,
    t.column_name,
    t.tag_value AS sensitivity_level,
    CONCAT(m.table_catalog, '.', m.table_schema, '.', m.mask_name) AS mask_function
FROM system.information_schema.column_tags t
LEFT JOIN system.information_schema.column_masks m
    ON t.catalog_name = m.table_catalog
   AND t.schema_name = m.table_schema
   AND t.table_name = m.table_name
   AND t.column_name = m.column_name
WHERE t.catalog_name = 'meridian_t12'
  AND t.tag_name = 'sensitivity_level';
```
